```python
collection.delete_many({})

collection.insert_many(data)

print("Documents inserted successfully")

print("Total documents:",
      collection.count_documents({}))
```

```python
failed_deliveries = collection.find(
    {"delivery_status": "Failed"}
)

for doc in failed_deliveries:
    print(
        doc["delivery_id"],
        doc["delivery_status"],
        doc["zone"]
    )
```

```python
collection.update_one(
    {"delivery_id": "DL00001"},
    {"$set": {"delivery_status": "Resolved"}}
)

updated_doc = collection.find_one(
    {"delivery_id": "DL00001"}
)

print(updated_doc["delivery_id"])

print(updated_doc["delivery_status"])
```

```python
test_record = {
    "delivery_id": "TEST001",
    "delivery_status": "Test",
    "zone": "Test Zone"
}

collection.insert_one(test_record)

print(
    "Before delete:",
    collection.count_documents(
        {"delivery_id": "TEST001"}
    )
)

collection.delete_one(
    {"delivery_id": "TEST001"}
)

print(
    "After delete:",
    collection.count_documents(
        {"delivery_id": "TEST001"}
    )
)
```

```python
collection.create_index(
    "delivery_status"
)

print("Indexes:")

for index in collection.list_indexes():
    print(index)
```
